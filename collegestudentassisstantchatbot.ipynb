```python
print("Total Questions:", len(df))

print("\nDataset columns:")
print(df.columns.tolist())

print("\nFirst 10 questions:")

for i, question in enumerate(df["question"].head(10), 1):
    print(i, question)
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")

print("Transformer model loaded successfully!")
```

```python
question_embeddings = model.encode(
    df["question"].tolist(),
    convert_to_numpy=True,
    show_progress_bar=True
)

print("Embeddings created successfully!")
print("Embedding shape:", question_embeddings.shape)
```

```python
def find_best_answer(user_question):

    user_embedding = model.encode(
        [user_question],
        convert_to_numpy=True
    )

    similarities = cosine_similarity(
        user_embedding,
        question_embeddings
    )[0]

    best_index = np.argmax(similarities)

    best_score = similarities[best_index]

    matched_question = df.iloc[best_index]["question"]

    answer = df.iloc[best_index]["answer"]

    return matched_question, answer, best_score
```

```python
CONFIDENCE_THRESHOLD = 0.45

def chatbot(user_question):

    user_question = user_question.strip()

    if not user_question:
        return "Please enter a question."

    matched_question, answer, score = find_best_answer(
        user_question
    )

    if score < CONFIDENCE_THRESHOLD:
        return (
            "Sorry, I couldn't find a reliable answer.\n\n"
            "You can ask me about:\n"
            "• College\n"
            "• Attendance\n"
            "• Exams\n"
            "• Assignments\n"
            "• Library\n"
            "• Projects\n"
            "• Internships\n"
            "• Placements\n"
            "• Student skills\n"
            "• Student activities"
        )

    return f"""
{answer}

------------------------------
Matched Question:
{matched_question}

Confidence: {score:.2f}
"""
```

```python
test_questions = [
    "Why is attendance important?",
    "How can I prepare for placements?",
    "What is an internship?",
    "Why are college projects important?"
]

for question in test_questions:

    print("=" * 60)
    print("QUESTION:", question)

    print(chatbot(question))
```

```python
def chat_function(message, history):

    if not message.strip():
        return "Please enter a question."

    return chatbot(message)


demo = gr.ChatInterface(
    fn=chat_function,
    title="🎓 College Student Assistant",
    description=(
        "Ask questions about college life, attendance, exams, "
        "assignments, projects, internships, placements and student skills."
    ),
    examples=[
        "Why is attendance important?",
        "How should I prepare for exams?",
        "What is an internship?",
        "How can I prepare for placements?",
        "Why are projects important?",
        "What are soft skills?"
    ]
)

demo.launch()
```
